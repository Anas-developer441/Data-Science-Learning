```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Style
sns.set(style="whitegrid")

# Q(1)-: Datewise Impact on the Stock

df = pd.read_csv("ai_financial_market_daily.csv")

# Convert Date column into datetime format
df['Date'] = pd.to_datetime(df['Date'], dayfirst=True)

# Extract Year
df['Year'] = df['Date'].dt.year

plt.figure(figsize=(14,6))

sns.lineplot(
    data=df,
    x='Date',
    y='Stock_Impact_%',
    hue='Company'
)

plt.title("Datewise Impact on the Stock")
plt.xlabel("Date")
plt.ylabel("Stock Impact %")

plt.xticks(rotation=45)

plt.show()
```

```python
#Q(2)-:  EXPENDITURE VS REVENUE YEAR-BY-YEAR

yearly_data = df.groupby('Year')[['R&D_Spending_USD_Mn',
                                  'AI_Revenue_USD_Mn']].sum().reset_index()

plt.figure(figsize=(12,6))

plt.plot(
    yearly_data['Year'],
    yearly_data['R&D_Spending_USD_Mn'],
    marker='o',
    label='R&D Expenditure'
)

plt.plot(
    yearly_data['Year'],
    yearly_data['AI_Revenue_USD_Mn'],
    marker='o',
    label='AI Revenue'
)

plt.title("Expenditure vs Revenue Year-by-Year")
plt.xlabel("Year")
plt.ylabel("Amount (USD Mn)")
plt.legend()
plt.show()
```

```python
#Q(3)-:EVENT IMPACT ANALYSIS
# Remove null events
event_df = df.dropna(subset=['Event'])

plt.figure(figsize=(12,6))

sns.boxplot(
    data=event_df,
    x='Event',
    y='Stock_Impact_%'
)

plt.title("Event Impact Analysis")
plt.xlabel("Event")
plt.ylabel("Stock Impact %")
plt.xticks(rotation=45)
plt.show()
```

```python
# Q(4)-: DAILY AVERAGE IMPACT ON THE STOCKS OF THE COMPANIES
daily_avg_rd = df.groupby('Company')['R&D_Spending_USD_Mn'].mean().reset_index()

plt.figure(figsize=(10,6))

sns.barplot(
    data=daily_avg_rd,
    x='Company',
    y='R&D_Spending_USD_Mn'
)

plt.title("Daily Average R&D Expenditure by Companies")
plt.xlabel("Company")
plt.ylabel("Average R&D Spending (USD Mn)")
plt.xticks(rotation=45)
plt.show()
```

```python
# Q(5)-: DAILY AVERAGE EXPENDITURE ON R&D BY THE COMPANIES


daily_avg_rd = df.groupby('Company')['R&D_Spending_USD_Mn'].mean().reset_index()

plt.figure(figsize=(10,6))

sns.barplot(
    data=daily_avg_rd,
    x='Company',
    y='R&D_Spending_USD_Mn'
)

plt.title("Daily Average R&D Expenditure by Companies")
plt.xlabel("Company")
plt.ylabel("Average R&D Spending (USD Mn)")
plt.xticks(rotation=45)
plt.show()
```

```python
# Q(6)-: MAXIMUM IMPACT % ON A COMPANY'S STOCKS


max_impact = df.groupby('Company')['Stock_Impact_%'].max().reset_index()

plt.figure(figsize=(10,6))

sns.barplot(
    data=max_impact,
    x='Company',
    y='Stock_Impact_%'
)

plt.title("Maximum Impact % on Company's Stocks")
plt.xlabel("Company")
plt.ylabel("Maximum Stock Impact %")
plt.xticks(rotation=45)
plt.show()

# Display values
print(max_impact)
```

```python
# =========================================================
# Q 7. MAXIMUM STOCK IMPACT WITH RESPECT TO YEAR & COMPANY
# =========================================================

year_company_max = df.groupby(
    ['Year', 'Company']
)['Stock_Impact_%'].max().reset_index()

plt.figure(figsize=(14,7))

sns.barplot(
    data=year_company_max,
    x='Year',
    y='Stock_Impact_%',
    hue='Company'
)

plt.title("Maximum Stock Impact wrt Year & Company")
plt.xlabel("Year")
plt.ylabel("Maximum Stock Impact %")
plt.legend(title='Company')
plt.show()

# Display table
print(year_company_max)
```
