### Question D1  
*7 marks - Hard*

The customer code has been entered five different ways across the files:

```
CUST-PK-1001        BR2019/PK-1002        Customer: PK-1003 (2020)
cust_pk_1004        2022-PK_1005
```

Some of these layouts also contain a **four-digit year**. A pattern of `(\d{4})` will therefore extract the wrong number from part of the file.

1. Write **one** regular expression that extracts the correct customer number from **all five** layouts into an integer column `cust_id`. Apply it to `transactions_2023`, `transactions_2024` and `customers`.
2. Prove your pattern was necessary: report **how many rows a naive `(\d{4})` would have got wrong**.

````python

import re

# 1. Regex pattern targeted specifically at the 4-digit ID following "PK-" or "PK_"
# Pattern Breakdown:
# (?i)          -> Case-insensitive matching
# PK[-_]        -> Matches 'PK-' or 'PK_' prefix
# (\d{4})       -> Captures EXACTLY the 4-digit customer number that follows PK




import re

# 1. Regex pattern: looks specifically for PK- or PK_ followed by 4 digits
pattern = r'(?i)PK[-_](\d{4})'

# Function to extract correct ID
def get_cust_id(val):
    match = re.search(pattern, str(val))
    return int(match.group(1)) if match else None

# Function to extract using naive regex
def get_naive_id(val):
    match = re.search(r'(\d{4})', str(val))
    return int(match.group(1)) if match else None

# 2. Apply to all three tables and count errors
total_wrong = 0

for df in [trans_2023, trans_2024, customers]:
    # Find column name (customer_id or customer_code)
    col = 'customer_code' if 'customer_code' in df.columns else 'customer_id'
    
    # Correct extraction
    df['cust_id'] = df[col].apply(get_cust_id)
    
    # Naive extraction
    naive_ids = df[col].apply(get_naive_id)
    
    # Count differences
    total_wrong += (df['cust_id'] != naive_ids).sum()

# 3. Print report
print(f"Total rows a naive (\\d{{4}}) got WRONG: {total_wrong}\n")

# Show sample results
customers[['customer_code', 'cust_id']].head()
````


### Question D2  
*5 marks - Hard*

Warehouse codes have been entered four ways:

```
WH-LHR-09        WH/ISB/04        wh-isb-05        WH ISB 08
```

1. Write **one** regular expression that pulls out the three-letter region code from all four, and store it as **uppercase** in a column `region_code` on both transaction files.
2. Report how many rows a stricter pattern of `WH-([A-Z]{3})-` would have **silently failed** to match.

````python
# Question D2 (Fixed Column Lookup)

import re

# 1. Flexible regex and strict regex
pattern = r'(?i)WH[/\s-]?([A-Za-z]{3})'
strict_pattern = r'WH-([A-Z]{3})-'

def get_region(val):
    match = re.search(pattern, str(val))
    return match.group(1).upper() if match else None

def get_strict_region(val):
    match = re.search(strict_pattern, str(val))
    return match.group(1) if match else None

# 2. Apply to both files safely
total_failed = 0

for df in [trans_2023, trans_2024]:
    # Dynamically find whichever column contains warehouse/wh info
    possible_cols = [c for c in df.columns if 'ware' in c.lower() or 'wh' in c.lower()]
    col = possible_cols[0]
    
    # Extract uppercase region code
    df['region_code'] = df[col].apply(get_region)
    
    # Count silent failures
    strict_res = df[col].apply(get_strict_region)
    total_failed += (df['region_code'].notna() & strict_res.isna()).sum()

# 3. Output results
print(f"Total rows strict pattern SILENTLY FAILED: {total_failed}\n")

print("Sample region_code extraction from 2023:")
print(trans_2023[['region_code']].head())
````

The `order_note` column contains bracketed status tags such as `[RUSH]`, `[pending]`, `[pdf]`.

Using **anchors**, report three separate counts for `transactions_2023`:

1. notes that **begin** with a tag
2. notes that **end** with a tag
3. notes that are **nothing but** a tag

Then show two example notes from each of the three groups.

````python
# Question D3

import pandas as pd

# 1. Standardize order_note to string, replacing missing values with empty strings
notes = trans_2023['order_note'].fillna('').astype(str).str.strip()

# 2. Define anchored regex patterns for bracketed tags
# Tag pattern: \[ [^\]]+ \] matches any bracketed text like [RUSH], [pending], [pdf]
tag_pat = r'\[[^\]]+\]'

# Anchors:
# ^  -> Start of string
# $  -> End of string
begins_pat = rf'^{tag_pat}'          # Starts with a tag
ends_pat   = rf'{tag_pat}$'          # Ends with a tag
exact_pat  = rf'^{tag_pat}$'         # Is strictly nothing but a tag

# 3. Create boolean masks
mask_exact  = notes.str.contains(exact_pat, regex=True)
mask_begins = notes.str.contains(begins_pat, regex=True) & ~mask_exact
mask_ends   = notes.str.contains(ends_pat, regex=True) & ~mask_exact

# 4. Report counts
print(f"Notes that begin with a tag (excluding exact matches): {mask_begins.sum()}")
print(f"Notes that end with a tag (excluding exact matches):   {mask_ends.sum()}")
print(f"Notes that are nothing but a tag:                   {mask_exact.sum()}\n")

# 5. Display two example notes from each group
print("--- Examples: Begins with a tag ---")
print(notes[mask_begins].head(2).tolist())

print("\n--- Examples: Ends with a tag ---")
print(notes[mask_ends].head(2).tolist())

print("\n--- Examples: Nothing but a tag ---")
print(notes[mask_exact].head(2).tolist())
````

````python
# Question D6

import warnings
import pandas as pd

warnings.filterwarnings('ignore', category=UserWarning)

# --- PART 1: URL Breakdown using Named Capturing Groups ---
# Pattern Breakdown:
# (?P<protocol>https?)   -> Captures http or https
# ://
# (?P<domain>[^/]+)      -> Captures domain name (everything up to next /)
# (?P<path>/.*)?         -> Captures optional path
url_pattern = r'(?P<protocol>https?)://(?P<domain>[^/]+)(?P<path>/.*)?'

# Find the URL column
url_col = [c for c in trans_2023.columns if 'url' in c.lower()][0]

# Extract named groups into a DataFrame
url_df = trans_2023[url_col].str.extract(url_pattern)

# Report results
print("--- Orders Count per Domain ---")
print(url_df['domain'].value_counts())

https_count = (url_df['protocol'] == 'https').sum()
print(f"\nTotal orders using HTTPS: {https_count}\n")


# --- PART 2: Standardizing Email Variations ---
# Find order note column
note_col = [c for c in trans_2023.columns if 'note' in c.lower()][0]

# Regex pattern matching e-mail, e mail, email, E-Mail
# e[- ]?mail matches 'e' followed by optional hyphen or space, then 'mail'
email_pattern = r'\be[- ]?mail\b'

# Find all spelling variations and their original counts
email_matches = trans_2023[note_col].dropna().str.extractall(f'({email_pattern})')[0]
print("--- Original Email Spellings & Counts ---")
print(email_matches.value_counts())

# Standardize all variations to 'email'
trans_2023[note_col] = trans_2023[note_col].str.replace(email_pattern, 'email', regex=True)

# Proof that it worked: count remaining variations
remaining_matches = trans_2023[note_col].dropna().str.extractall(f'({email_pattern})')[0]
print("\n--- Proof of Standardization (Post-replacement Counts) ---")
print(remaining_matches.value_counts())
````

````python
# F1 - Revenue and Sales Representative Analysis
# the logic 
#Gross Revenue = Quantity × Unit Price

#Discount Amount = Gross Revenue × Discount Rate

#Net Revenue = Gross Revenue − Discount Amount




# Convert quantity to numeric
m['qty'] = pd.to_numeric(m['qty'], errors='coerce')

# Convert unit price to numeric
# Removes "PKR" and commas if they are present
m['unit_price'] = (
    m['unit_price']
    .astype(str)
    .str.replace('PKR', '', regex=False)
    .str.replace(',', '', regex=False)
    .str.strip()
)

m['unit_price'] = pd.to_numeric(
    m['unit_price'],
    errors='coerce'
)


# 1. Calculate gross revenue
m['gross_revenue'] = m['qty'] * m['unit_price']


# Convert discount percentage to numeric
m['discount_rate'] = (
    m['discount_pct']
    .astype(str)
    .str.replace('%', '', regex=False)
)

m['discount_rate'] = pd.to_numeric(
    m['discount_rate'],
    errors='coerce'
).fillna(0)

# Convert percentage to decimal
m['discount_rate'] = m['discount_rate'] / 100


# Calculate discount amount
m['discount_amount'] = (
    m['gross_revenue'] * m['discount_rate']
)

# Calculate net revenue
m['net_revenue'] = (
    m['gross_revenue'] - m['discount_amount']
)


# Report totals
print("Total Net Revenue:    PKR {:,.2f}".format(
    m['net_revenue'].sum()
))

print("Total Discount Given: PKR {:,.2f}".format(
    m['discount_amount'].sum()
))


# 2. Number of orders and total net revenue per sales rep
rep_summary = (
    m.groupby('sales_rep')
     .agg(
         order_count=('order_id', 'count'),
         total_net_revenue=('net_revenue', 'sum')
     )
     .reset_index()
)

print("\n--- Sales Rep Performance Summary ---")
print(rep_summary)


# 3. Use get_group to show first 3 transactions of one rep
rep_groups = m.groupby('sales_rep')

sample_rep = rep_summary.iloc[0]['sales_rep']

print(f"\n--- First 3 Transactions for Rep: {sample_rep} ---")

print(
    rep_groups.get_group(sample_rep)[
        ['order_id', 'qty', 'unit_price',
         'gross_revenue', 'net_revenue']
    ].head(3)
)
````

1. Build a pivot table with **region name** down the side, **category** across the top, and **total** net revenue in the cells.
2. For each region, find the single **brand** that earned the most net revenue, and how much.

For part 2 you must use a **loop over the groups** and track the maximum yourself - do not use `idxmax`, `max`, `nlargest` or `sort_values`.

````python
import pandas as pd

# Column names
region_col = 'region_name'
category_col = 'category'
net_revenue_col = 'net_revenue'

# Create brand from the first word of product_details
m['brand'] = m['product_details'].str.split().str[0]


# ============================================================
# 1. Pivot Table: Region vs Category - Total Net Revenue
# ============================================================

pivot_revenue = pd.pivot_table(
    m,
    values=net_revenue_col,
    index=region_col,
    columns=category_col,
    aggfunc='sum',
    fill_value=0
)

print("--- Net Revenue Pivot Table (Region vs Category) ---")
print(pivot_revenue)

print("\n" + "="*60 + "\n")


# ============================================================
# 2. Top Brand per Region - Manual Loop
# ============================================================

region_brand_grouped = m.groupby(
    [region_col, 'brand']
)['net_revenue'].sum()

print("--- Top Earning Brand per Region ---")

for region in m[region_col].unique():

    if pd.isna(region):
        continue

    top_brand = None
    max_revenue = -1

    # Loop through all region-brand groups
    for (r, b), total_revenue in region_brand_grouped.items():

        if r == region:

            if total_revenue > max_revenue:
                max_revenue = total_revenue
                top_brand = b

    print(
        f"Region: {region:<15} | "
        f"Top Brand: {top_brand:<15} | "
        f"Net Revenue: ${max_revenue:,.2f}"
    )
````
