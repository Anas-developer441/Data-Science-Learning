```python
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns 

df=pd.read_csv('HR_Data_MNC_Data Science Lovers.csv')

df.head()
```

```python
# Question 1
# Q1  What percentage of employees are currently “Active” according to the pie chart?
# Import libraries


# Count employee status
status_counts = df['Status'].value_counts()

# Calculate percentages
status_percentage = (status_counts / len(df)) * 100

# Print percentages
print("Employee Status Percentage:\n")
print(status_percentage)

# Get Active percentage only
active_percentage = status_percentage['Active']

print(f"\nQ1a Answer: {active_percentage:.2f}% employees are currently Active.")

# Plot pie chart
plt.figure(figsize=(7,7))

plt.pie(
    status_counts,
    labels=status_counts.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title("Employee Status Distribution")

plt.show()
```

```python
# QUESTION 2

# Count each employee status
status_counts = df['Status'].value_counts()

# Convert into percentages
status_percentage = (status_counts / len(df)) * 100

# Round percentages to 1 decimal place
status_percentage = status_percentage.round(1)

# Display all percentages
print("Employee Status Percentages:\n")
print(status_percentage)

# Find categories with exactly 5.0%
exact_five = status_percentage[status_percentage == 5.0]

print("\nCategories with exactly 5.0%:\n")
print(exact_five)

# If exactly two categories exist
if len(exact_five) == 2:
    categories = list(exact_five.index)
    
    print(f"\nQ1b Answer: {categories[0]} and {categories[1]}")
```

```python
#QUESTION -> 3 

# Count employees in each department
department_counts = df['Department'].value_counts()

# Display employee count per department
print("Employee Count by Department:\n")
print(department_counts)

# Find department with highest employees
top_department = department_counts.idxmax()
top_count = department_counts.max()

print(f"\nDepartment with the most employees: {top_department}")
print(f"Total Employees: {top_count}")

# Plot bar chart
plt.figure(figsize=(10,6))

department_counts.plot(kind='bar')

plt.title("Employee Count by Department")
plt.xlabel("Department")
plt.ylabel("Number of Employees")
plt.xticks(rotation=45)

plt.show()
```

```python
#QUESTION -> 4 
# Count employees in each department
department_counts = df['Department'].value_counts()

# Display all department counts
print("Employee Count by Department:\n")
print(department_counts)

# Find departments with fewer than 200,000 employees
below_threshold = department_counts[department_counts < 200000]

print("\nDepartments with fewer than 200,000 employees:\n")
print(below_threshold)

# If you only want the department names
print("\nDepartment Names:")
print(list(below_threshold.index))

# Bar chart for visualization
plt.figure(figsize=(10,6))

department_counts.plot(kind='bar')

# Threshold line
plt.axhline(y=200000, linestyle='--')

plt.title("Department Employee Count")
plt.xlabel("Department")
plt.ylabel("Number of Employees")
plt.xticks(rotation=45)

plt.show()
```

```python
# QUESTION NUMBER 5


# Count employees in each department
department_counts = df['Department'].value_counts()

# Display all department counts
print("Employee Count by Department:\n")
print(department_counts)

# Get IT department employee count
it_count = department_counts['IT']

# Calculate 50% of IT department
half_it = it_count * 0.5

print(f"\nIT Department Count: {it_count}")
print(f"50% of IT Department: {half_it}")

# Find department closest to 50% of IT count
# Excluding IT itself
other_departments = department_counts.drop('IT')

closest_department = (other_departments - half_it).abs().idxmin()
closest_count = other_departments[closest_department]

print(f"\nDepartment closest to 50% of IT headcount: {closest_department}")
print(f"Employee Count: {closest_count}")

# Visualization
plt.figure(figsize=(10,6))

department_counts.plot(kind='bar')

# Show 50% IT threshold line
plt.axhline(y=half_it, linestyle='--')

plt.title("Department Employee Counts")
plt.xlabel("Department")
plt.ylabel("Number of Employees")
plt.xticks(rotation=45)

plt.show()
```

```python

# QUESTION NO 6


# Calculate average salary by department
avg_salary = df.groupby('Department')['Salary_INR'].mean().sort_values(ascending=False)

# Display average salaries
print("Average Salary by Department:\n")
print(avg_salary)

# Find department with highest average salary
highest_department = avg_salary.idxmax()
highest_salary = avg_salary.max()

print(f"\nDepartment with Highest Average Salary: {highest_department}")
print(f"Average Salary: {highest_salary:.2f} INR")

# Visualization
plt.figure(figsize=(10,6))

avg_salary.plot(kind='bar')

plt.title("Average Salary by Department")
plt.xlabel("Department")
plt.ylabel("Average Salary (INR)")
plt.xticks(rotation=45)

plt.show()
```

```python

# Calculate average salary by department
avg_salary = df.groupby('Department')['Salary_INR'].mean()

# Get Finance and IT salaries
finance_salary = avg_salary['Finance']
it_salary = avg_salary['IT']

# Calculate difference
difference = abs(it_salary - finance_salary)

# Display results
print(f"Finance Average Salary: {finance_salary:.2f}")
print(f"IT Average Salary: {it_salary:.2f}")

print(f"\nApproximate Salary Difference: {difference:.2f}")
```

```python
# QUESTION NO 8
# Calculate average salary by department
avg_salary = df.groupby('Department')['Salary_INR'].mean()

# Target salary
target_salary = 800000

# Find closest departments
closest_departments = (avg_salary - target_salary).abs().sort_values().head(2)

# Display results
print("Departments closest to 800,000 average salary:\n")

for dept, diff in closest_departments.items():
    print(f"{dept} -> Average Salary: {avg_salary[dept]:.2f}")
```

```python
# QUESTION NO 9 

# Calculate average salary by job title
job_salary = df.groupby('Job_Title')['Salary_INR'].mean().sort_values(ascending=False)

# Display salaries
print("Average Salary by Job Title:\n")
print(job_salary)

# Highest paid role
highest_role = job_salary.idxmax()
highest_salary = job_salary.max()

print(f"\nHighest Paying Role: {highest_role}")
print(f"Average Salary: {highest_salary:.2f}")

# Visualization
plt.figure(figsize=(12,6))

job_salary.plot(kind='bar')

plt.title("Average Salary by Job Title")
plt.xlabel("Job Title")
plt.ylabel("Average Salary")
plt.xticks(rotation=45)

plt.show()
```
