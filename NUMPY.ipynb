# MATRIX MULTIPLCATION 
* ->.dot(A+b) (element wise )example
@-> matrix to matrix calculation


1. * — Element-wise Multiplication (Hadamard Product)The * operator multiplies corresponding elements in the exact same positions.
    How it works:If you have two matrices $A$ and $B$:$$A = \begin{bmatrix} 1 & 2 \\ 3 & 4 \end{bmatrix}, \quad B = \begin{bmatrix} 5 & 6 \\ 7 & 8 \end{bmatrix}$$Applying A * B multiplies $A[i, j]$ by $B[i, j]$:$$A * B = \begin{bmatrix} 1 \times 5 & 2 \times 6 \\ 3 \times 7 & 4 \times 8 \end{bmatrix} = \begin{bmatrix} 5 & 12 \\ 21 & 32 \end{bmatrix}




2. @ — Matrix Multiplication (Dot Product)

3. The @ operator performs standard Linear Algebra matrix multiplication (taking the dot product of rows from $A$ and columns from $B$).How it works:Using the same matrices $A$ and $B$:Top-Left: Row 1 of $A$ $\cdot$ Column 1 of $B$ $\rightarrow (1 \times 5) + (2 \times 7) = 5 + 14 = \mathbf{19}$Top-Right: Row 1 of $A$ $\cdot$ Column 2 of $B$ $\rightarrow (1 \times 6) + (2 \times 8) = 6 + 16 = \mathbf{22}$Bottom-Left: Row 2 of $A$ $\cdot$ Column 1 of $B$ $\rightarrow (3 \times 5) + (4 \times 7) = 15 + 28 = \mathbf{43}$Bottom-Right: Row 2 of $A$ $\cdot$ Column 2 of $B$ $\rightarrow (3 \times 6) + (4 \times 8) = 18 + 32 = \mathbf{50}$$$A \text{ @ } B = \begin{bmatrix} 19 & 22 \\ 43 & 50 \end{bmatrix}$$
